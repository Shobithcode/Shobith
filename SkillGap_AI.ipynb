# SkillGap AI – AI-Based Skill Gap Analyzer

An AI/ML-based system that analyzes a user's skills, identifies skill gaps, predicts suitable job roles, and generates a personalized learning roadmap.

## 1. Job Role & Skill Database

```python
# SkillGap AI
# AI-Based Skill Gap Analyzer

print("======================================")
print("        SKILLGAP AI")
print("   AI-Based Skill Gap Analyzer")
print("======================================")

# Job roles and required skills
job_roles = {
    "AI/ML Intern": [
        "python",
        "machine learning",
        "numpy",
        "pandas",
        "scikit-learn",
        "statistics"
    ],

    "Data Analyst": [
        "python",
        "pandas",
        "sql",
        "excel",
        "statistics",
        "data visualization"
    ],

    "Python Developer": [
        "python",
        "oops",
        "git",
        "sql",
        "apis",
        "django"
    ]
}

# Select role
print("\nAvailable Job Roles:")
for role in job_roles:
    print("-", role)

role = input("\nEnter your target job role: ")

if role not in job_roles:
    print("Invalid role. Please choose one from the list.")
else:
    # Get user's skills
    skills_input = input(
        "\nEnter your current skills separated by commas: "
    )

    user_skills = set(
        skill.strip().lower()
        for skill in skills_input.split(",")
    )

    required_skills = set(job_roles[role])

    # Find matching and missing skills
    matching_skills = user_skills.intersection(required_skills)
    missing_skills = required_skills - user_skills

    # Calculate match percentage
    match_percentage = (
        len(matching_skills) / len(required_skills)
    ) * 100

    print("\n======================================")
    print("          ANALYSIS RESULT")
    print("======================================")

    print(f"\nTarget Role: {role}")
    print(f"Skill Match: {match_percentage:.2f}%")

    print("\n✅ Matching Skills:")
    for skill in matching_skills:
        print("-", skill.title())

    print("\n❌ Missing Skills:")
    for skill in missing_skills:
        print("-", skill.title())

    print("\n📚 Suggested Learning Roadmap:")

    for skill in missing_skills:
        print(f"Learn → {skill.title()}")

    print("\n======================================")
    print("        Analysis Completed!")
    print("======================================")
```

```python
import matplotlib.pyplot as plt

# Calculate skill percentages
matched_count = len(matching_skills)
missing_count = len(missing_skills)

labels = ["Matched Skills", "Missing Skills"]
values = [matched_count, missing_count]

plt.figure(figsize=(7, 5))
plt.bar(labels, values)

plt.title(f"Skill Gap Analysis - {role}")
plt.ylabel("Number of Skills")

for i, value in enumerate(values):
    plt.text(i, value + 0.1, str(value), ha="center")

plt.show()
```

## 2. NLP-Based Skill Extraction

```python
# NLP-based Skill Extraction

import re

def extract_skills(resume_text):
    skill_database = [
        "python",
        "java",
        "c",
        "machine learning",
        "deep learning",
        "numpy",
        "pandas",
        "scikit-learn",
        "sql",
        "excel",
        "statistics",
        "data visualization",
        "git",
        "github",
        "django",
        "apis",
        "html",
        "css",
        "javascript"
    ]

    resume_text = resume_text.lower()

    found_skills = []

    for skill in skill_database:
        if re.search(r'\b' + re.escape(skill) + r'\b', resume_text):
            found_skills.append(skill)

    return found_skills


# Example resume text
resume_text = """
I am an AIML engineering student.
I have experience with Python, Pandas, NumPy,
Machine Learning, Git and GitHub.
I am currently learning SQL and Data Visualization.
"""

extracted_skills = extract_skills(resume_text)

print("🤖 NLP SKILL EXTRACTION")
print("========================")

print("\nSkills detected from resume:")

for skill in extracted_skills:
    print("✓", skill.title())
```

## 3. Skill Gap Analysis

```python
# Skill Gap Analysis using NLP-extracted skills

def analyze_skill_gap(extracted_skills, target_role):
    extracted_skills = set(skill.lower() for skill in extracted_skills)
    required_skills = set(job_roles[target_role])

    matched = extracted_skills.intersection(required_skills)
    missing = required_skills - extracted_skills

    score = (len(matched) / len(required_skills)) * 100

    print("======================================")
    print("       AI SKILL GAP ANALYSIS")
    print("======================================")

    print(f"\nTarget Role: {target_role}")
    print(f"Skill Match Score: {score:.2f}%")

    print("\n✅ Matched Skills:")
    for skill in sorted(matched):
        print("-", skill.title())

    print("\n❌ Missing Skills:")
    for skill in sorted(missing):
        print("-", skill.title())

    print("\n📚 Recommended Learning:")
    for skill in sorted(missing):
        print(f"Learn → {skill.title()}")

    return score


# Analyze the sample resume
score = analyze_skill_gap(
    extracted_skills,
    "AI/ML Intern"
)
```

## 4. Machine Learning Job Role Prediction

```python
# Machine Learning: Job Role Prediction

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression

# Training examples
training_texts = [
    "python machine learning numpy pandas scikit-learn statistics",
    "python deep learning numpy pandas machine learning",
    "python pandas sql excel statistics data visualization",
    "sql excel pandas data visualization statistics",
    "python oops git sql apis django",
    "python git django apis sql",
]

training_labels = [
    "AI/ML Intern",
    "AI/ML Intern",
    "Data Analyst",
    "Data Analyst",
    "Python Developer",
    "Python Developer"
]

# Convert text into numerical features
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(training_texts)

# Train ML model
model = LogisticRegression()
model.fit(X, training_labels)

# Convert extracted skills into text
resume_for_prediction = " ".join(extracted_skills)

# Predict suitable job role
resume_vector = vectorizer.transform([resume_for_prediction])
predicted_role = model.predict(resume_vector)[0]

print("======================================")
print("       ML JOB ROLE PREDICTION")
print("======================================")

print("\nDetected Skills:")
print(", ".join(skill.title() for skill in extracted_skills))

print("\n🤖 Predicted Suitable Role:")
print(predicted_role)
```

## 5. Interactive Skill Analyzer

```python
# Interactive SkillGap AI Analyzer

print("======================================")
print("        SKILLGAP AI ANALYZER")
print("======================================")

user_resume = input("\nPaste your resume/skills here: ")

# Extract skills from user input
user_extracted_skills = extract_skills(user_resume)

print("\n🔍 Skills Detected:")
if user_extracted_skills:
    for skill in user_extracted_skills:
        print("✓", skill.title())
else:
    print("No matching skills detected.")

# Predict suitable role
user_text = " ".join(user_extracted_skills)
user_vector = vectorizer.transform([user_text])

predicted_role = model.predict(user_vector)[0]

print("\n🤖 Predicted Suitable Role:")
print(predicted_role)

# Analyze skill gap
if predicted_role in job_roles:
    required = set(job_roles[predicted_role])
    detected = set(user_extracted_skills)

    matched = detected.intersection(required)
    missing = required - detected

    score = (len(matched) / len(required)) * 100

    print(f"\n📊 Skill Match Score: {score:.2f}%")

    print("\n❌ Skills to Learn:")
    for skill in sorted(missing):
        print("→", skill.title())

print("\n======================================")
print("        Analysis Completed!")
print("======================================")
```

## 6. Personalized Learning Roadmap

```python
# Personalized Learning Roadmap

print("\n======================================")
print("     PERSONALIZED LEARNING ROADMAP")
print("======================================")

priority_skills = {
    "python": "High",
    "machine learning": "High",
    "numpy": "Medium",
    "pandas": "Medium",
    "scikit-learn": "High",
    "statistics": "High",
    "sql": "High",
    "excel": "Medium",
    "data visualization": "Medium",
    "git": "Medium",
    "oops": "High",
    "apis": "Medium",
    "django": "Medium"
}

print("\n📚 Recommended Skills to Learn:")

for skill in sorted(missing):
    priority = priority_skills.get(skill.lower(), "Medium")
    print(f"→ {skill.title()} | Priority: {priority}")

print("\n💡 Suggested Learning Order:")

high_priority = [
    skill for skill in missing
    if priority_skills.get(skill.lower(), "Medium") == "High"
]

for i, skill in enumerate(high_priority, 1):
    print(f"{i}. {skill.title()}")

print("\n======================================")
print("        Roadmap Generated!")
print("======================================")
```
